```python
import zipfile, os, pathlib
zip_path="/content/task3Images.zip"
zf=zipfile.ZipFile(zip_path)
names=zf.namelist()
len(names), names[:30]
```

```python
from PIL import Image, ImageDraw
import io, math
imgs=[]
for n in names:
    if n.lower().endswith(('.jpg','.jpeg','.png')):
        im=Image.open(io.BytesIO(zf.read(n))).convert("RGB")
        imgs.append((n,im.copy()))
[(n,im.size) for n,im in imgs[:5]], len(imgs)
```

```python
display(imgs[0][1].resize((1024,528)))
```

```python
# contact sheet of all images, thumbnails
thumb_w, thumb_h = 256, 132
sheet=Image.new("RGB",(thumb_w*3, (thumb_h+20)*5),"white")
d=ImageDraw.Draw(sheet)
for i,(n,im) in enumerate(imgs[:15]):
    x=(i%3)*thumb_w; y=(i//3)*(thumb_h+20)
    sheet.paste(im.resize((thumb_w,thumb_h)),(x,y))
    d.text((x+4,y+2),os.path.basename(n),fill="yellow",stroke_width=1,stroke_fill="black")
display(sheet)
```

```python
# display sensor crops from samples with high res
for idx in [0,1,2,3,4,5,10,20,30,44]:
    n,im=imgs[idx]
    print(n)
    display(im.crop((256,0,512,264)).resize((768,792)))
```
